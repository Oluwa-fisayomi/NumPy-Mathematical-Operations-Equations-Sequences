## Business Analytics — Sales Performance Calculator

### Scenario

A growing retail business recorded its daily sales for one week.

The goal is to use NumPy to analyze the business's weekly sales performance.

```python
import numpy as np
```

## Task 1 - Create the dataset

```python
daily_sales = np.array([125000, 150000, 175000, 140000, 190000, 210000, 160000])

print(f"""
Daily Sales = {daily_sales}
Type = {type(daily_sales)}
Shape = {daily_sales.shape}
""")
```

## Explanation

The dataset contains 7 values because the business recorded sales for 7 days.

The data is stored as a NumPy array. Its shape is (7,), meaning it is a one-dimensional array containing 7 elements.

## Task 2 - Calculate Total Sales

```python
total_sales = np.sum(daily_sales)

print(f"Total Sales: {total_sales}")
```

## Explanation



## Task 3 - Calculate Average Daily Sales

```python
average_sales = np.mean(daily_sales)

print(f"Average Daily Sales = {average_sales:.2f}")
```

## Task 4 - Apply a Sales Adjustment

```python
adjusted_sales = daily_sales * 1.10

print(f"Adjusted Sales: {adjusted_sales}")
```

## Explanation

The entire array was multiplied by 1.10.

This represents the original sales plus a 10% increase for each day.

## Task 5 - Compare Original and Adjusted Sales

```python
additional_revenue = adjusted_sales - daily_sales

print(f"Additional Sales: {additional_revenue}")
```

## Explanation

The additional revenue represents the amount each day's sales would increase by under the hypothetical 10% increase.

For example, the first day's original sales were ₦125,000.

A 10% increase gives:

₦125,000 × 10% = ₦12,500

Therefore, the additional revenue for that day is ₦12,500.

## Task 6 - Use NumPy's Mathematical Functions

```python
average_sales_check = np.mean(daily_sales)

print(f"Total Sales: {total_sales}")
print(f"Average Daily Sales: {average_sales_check:.2f}")
```

## Explanation


The total sales figure tells the business how much revenue was generated across the entire week. The average daily sales figure tells the business how much revenue was generated on a typical day.

The total is useful for understanding overall weekly performance, while the average makes it easier to understand daily performance and compare this week with other periods.

## Business Thinking Challenge


The business generated total weekly sales of ₦1,150,000, with an average daily sales value of approximately ₦164,285.71. This shows the overall revenue generated during the week as well as the typical amount generated per day. The business can use these figures to monitor sales performance and compare future weeks.

# HANDS-ON EXERCISE 2
## Education Analytics — Student Performance Analysis

### Scenario

A training program recorded the scores of a group of students.

The goal is to analyze the students' overall performance and determine how individual scores differ from the group average.

## Task 1 - Create and Inspect the Data

```python
import numpy as np
```

```python
scores = np.array([62, 75, 81, 69, 88, 94, 73, 85, 77, 91])

print(f"""
Scores: {scores}
Type: {type(scores)}
Shape: {scores.shape}
""")
```

## Task 2 - Calculate the Mean Score

```python
mean_score = np.mean(scores)

print(f"Mean Score: {mean_score}")
```

## Task 3 - Calculate Each Student's Deviation

```python
deviations = scores - mean_score

print(f"Deviations: {deviations}")
```

## Explanation

Each deviation shows how far a student's score is from the group average of 79.5.

A negative value means the score is below the average, while a positive value means the score is above the average.

## Task 4 - Identify Positive and Negative Deviations

## Explanation

A positive deviation means that the student's score is above the group average.

A negative deviation means that the student's score is below the group average.

A value close to zero means that the student's score is close to the group average.

For example, the score of 94 has a deviation of +14.5, meaning it is 14.5 points above the average. The score of 62 has a deviation of -17.5, meaning it is 17.5 points below the average.

## Task 5 - Square the Deviations

```python
squared_deviations = deviations ** 2

print(f"Squared Deviations: {squared_deviations}")
```

## Explanation


Squaring a number makes a negative value positive because a negative number multiplied by another negative number produces a positive result.

For example:

(-17.5)² = 306.25

This allows the deviations to be combined without positive and negative values cancelling each other out.

## Task 6 - Calculate the sum of squared deviations

```python
sum_squared_deviations = np.sum(squared_deviations)

print(f"Sum of the Squared Deviations: {sum_squared_deviations}")
```

## Task 7 - Connect the Steps to Standard Deviation

## Connecting the Steps to Standard Deviation

The process of calculating standard deviation can be understood through these stages:

1. Calculate the mean.
2. Calculate the deviations of each value from the mean.
3. Square the deviations.
4. Sum the squared deviations.
5. Continue with the appropriate division.
6. Take the square root.

In this exercise:

- Mean = 79.5
- Sum of squared deviations = 932.5

Standard deviation is useful because it helps show how spread out the student scores are around the average. While the mean tells us the typical score, standard deviation gives additional information about the variation in the scores.

## Industry Thinking Challenge


The average score gives a general idea of the group's performance, but it does not show how individual students performed relative to that average. Looking at deviations helps identify students whose scores are considerably above or below the average. This can help the program coordinator understand differences in student performance and identify areas that may require additional attention.

## Engineering & Scientific Computing — Trigonometric Series and Convergence

### Scenario

An engineering/scientific computing team wants to understand how a mathematical approximation changes as more terms are included.

We will use NumPy to create a sequence, calculate a trigonometric component, generate series terms, and observe convergence.

## Task 1 - Define the Angle

```python
theta = 30

theta_radians = np.radians(theta)

print(f"Angle in Degrees: {theta}")
print(f"Angle in Radians: {theta_radians:.2f}")
```

## Task 2 - Create the Sequence of Terms

```python
k = np.arange(1, 100, 1)

print(f"First Value: {k[0]}")
print(f"Last Value: {k[-1]}")
print(f"Number of Values Generated: {len(k)}")
```

## Explanation

The sequence starts at 1 and stops before 100.

Because the stop value in np.arange() is not included, the last value is 99.

Therefore, the sequence contains 99 values.

## Task 3 - Calculate the Trigonometric Component

```python
sin_theta = np.sin(theta_radians)

print(f"sin(theta) = {sin_theta:.4f}")
```

## Task 4 - Build the Series Terms

```python
series_terms = sin_theta / k

print(f"Series Terms: {series_terms}")
```

## Explanation

NumPy performs element-wise division.

The single value sin(theta) is divided by every value in the k array.

Therefore, the operation produces one series term for each value of k.

## Task 5 - Calculate the Sum

```python
series_sum = np.sum(series_terms)

print(f"Sum of the Generated Terms: {series_sum:.3f}")
```

## Task 6 - Increase the Number of Terms

```python
k_first = np.arange(1, 100, 1)
terms_first = sin_theta / k_first
print(f"Sum of the First Terms: {np.sum(terms_first)}")

k_second = np.arange(1, 1000, 1)
terms_second = sin_theta / k_second
print(f"Sum of the Second Terms: {np.sum(terms_second)}")

k_third = np.arange(1, 10000, 1)
terms_third = sin_theta / k_third
print(f"Sum of the Third Terms: {np.sum(terms_third)}")
```

## Task 7 - Investigate Convergence

## Convergence Results

| Number of Terms | Series Result |
|---:|---:|
| 99 | 2.5886887588 |
| 999 | 3.7422354303 |
| 9,999 | 4.8937530180 |

As the number of terms increases, the series result also increases. The approximation changes because more terms are being added to the series.

## Task 8 - Explain Convergence

## Explanation of Convergence

### 1. What happened when fewer terms were used?

When fewer terms were used, the calculated series result was smaller because fewer terms were included in the calculation.

### 2. What happened when the number of terms increased?

When the number of terms increased from 99 to 999 and then to 9,999, the series result increased.

### 3. Why does increasing the number of terms change the approximation?

Increasing the number of terms adds more values to the calculation. Since the additional terms contribute to the total, the resulting approximation changes.

### 4. What does it mean when the result gets closer to a limiting value?

It means that the calculation is approaching a value that the series tends toward as more terms are included. This behavior is known as convergence.

## Sports Performance Analysis

### Scenario

A football player records the number of successful passes made in five matches.

The goal is to use NumPy to analyze the player's passing performance and calculate the total, average, improvement, and percentage-related information.

### Objective

The objective is to use NumPy mathematical operations to analyze the player's performance across five matches.

```python
matches = np.array([32, 38, 35, 42, 45])

print(f"Successful Passes: {matches}")
```

## Bonus Calculation 1 - Total Successful Passes

```python
total_passes = np.sum(matches)

print(f"Total Successful Passes: {total_passes}")
```

## Bonus Calculation 2 - Average Successful Passes

```python
average_passes = np.mean(matches)

print(f"Average Successful Passes: {average_passes}")
```

## Bonus Calculation 3 - Difference from Average

```python
pass_deviations = matches - average_passes

print(f"Deviations from Average: {pass_deviations}")
```

## Bonus Calculation 4 - 10% Performance Increase

```python
projected_passes = matches * 1.10

print(f"Projected Passes After 10% Increase: {projected_passes}")
```

## Bonus Calculation 5 - Difference Between Projected and Original

```python
additional_passes = projected_passes - matches

print(f"Additional Projected Passes: {additional_passes}")
```

## Results

The player recorded 192 successful passes across five matches.

The average number of successful passes per match was 38.4.

The deviations show how each match compared with the average. The first match was 6.4 passes below the average, while the fifth match was 6.6 passes above the average.

If the player's successful passes increased by 10%, the projected performances would range from 35.2 to 49.5 successful passes per match.

## Interpretation

The results show that the player's passing performance varied across the five matches. The average of 38.4 provides a general measure of performance, while the deviations show how individual matches differed from that average.

The 10% projection also demonstrates how NumPy can be used to apply the same mathematical operation to an entire dataset.
